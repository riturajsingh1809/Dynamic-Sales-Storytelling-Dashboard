```python
import pandas as pd
import numpy as np

# 1. Load Cleaned Dataset
df = pd.read_excel("Cleaned_Dataset.xlsx")

print("Dataset Loaded Successfully")
print("Shape:", df.shape)

# 2. Display First 5 Rows
print("\nFirst 5 Rows:")
display(df.head())

# 3. Dataset Information
print("\nDataset Information:")
df.info()

# 4. Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Check Duplicate Rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# 6. Statistical Summary
print("\nStatistical Summary:")
display(df.describe())

# 7. Total Sales
total_sales = df["sales"].sum()

print("\nTotal Sales:", total_sales)

# 8. Total Profit
total_profit = df["profit"].sum()

print("Total Profit:", total_profit)

# 9. Total Quantity
total_quantity = df["quantity"].sum()

print("Total Quantity:", total_quantity)

# 10. Total Orders
total_orders = df["order_id"].nunique()

print("Total Orders:", total_orders)

# 11. Average Sales
average_sales = df["sales"].mean()

print("Average Sales:", average_sales)

# 12. Average Profit
average_profit = df["profit"].mean()

print("Average Profit:", average_profit)

# 13. Profit Margin
profit_margin = (
    total_profit / total_sales
) * 100

print(
    "Profit Margin:",
    round(profit_margin, 2),
    "%"
)

# 14. Region-wise Analysis
region_analysis = (
    df.groupby("region")
    .agg(
        Total_Sales=("sales", "sum"),
        Total_Profit=("profit", "sum"),
        Total_Quantity=("quantity", "sum"),
        Average_Sales=("sales", "mean")
    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\nRegion Analysis:")
display(region_analysis)

# 15. Category-wise Analysis
category_analysis = (
    df.groupby("category")
    .agg(
        Total_Sales=("sales", "sum"),
        Total_Profit=("profit", "sum"),
        Total_Quantity=("quantity", "sum"),
        Average_Sales=("sales", "mean")
    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\nCategory Analysis:")
display(category_analysis)

# 16. Product-wise Analysis
product_analysis = (
    df.groupby("product")
    .agg(
        Total_Sales=("sales", "sum"),
        Total_Profit=("profit", "sum"),
        Total_Quantity=("quantity", "sum"),
        Average_Sales=("sales", "mean")
    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\nProduct Analysis:")
display(product_analysis)

# 17. Customer Segment Analysis
segment_analysis = (
    df.groupby("customer_segment")
    .agg(
        Total_Sales=("sales", "sum"),
        Total_Profit=("profit", "sum"),
        Total_Quantity=("quantity", "sum"),
        Average_Sales=("sales", "mean")
    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\nCustomer Segment Analysis:")
display(segment_analysis)

# 18. Monthly Sales
monthly_sales = (
    df.groupby(
        df["order_date"].dt.to_period("M")
    )["sales"]
    .sum()
    .reset_index()
)

monthly_sales["order_date"] = (
    monthly_sales["order_date"].astype(str)
)

print("\nMonthly Sales:")
display(monthly_sales)

# 19. Monthly Profit
monthly_profit = (
    df.groupby(
        df["order_date"].dt.to_period("M")
    )["profit"]
    .sum()
    .reset_index()
)

monthly_profit["order_date"] = (
    monthly_profit["order_date"].astype(str)
)

print("\nMonthly Profit:")
display(monthly_profit)

# 20. Yearly Sales
yearly_sales = (
    df.groupby(
        df["order_date"].dt.year
    )["sales"]
    .sum()
    .reset_index()
)

yearly_sales.columns = [
    "Year",
    "Sales"
]

print("\nYearly Sales:")
display(yearly_sales)

# 21. Yearly Profit
yearly_profit = (
    df.groupby(
        df["order_date"].dt.year
    )["profit"]
    .sum()
    .reset_index()
)

yearly_profit.columns = [
    "Year",
    "Profit"
]

print("\nYearly Profit:")
display(yearly_profit)

# 22. Region and Category Analysis
region_category = (
    df.groupby(
        ["region", "category"]
    )
    .agg(
        Total_Sales=("sales", "sum"),
        Total_Profit=("profit", "sum")
    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\nRegion and Category Analysis:")
display(region_category)

# 23. Top Product
top_product = product_analysis.iloc[0]["product"]

print("\nTop Product:", top_product)

# 24. Top Region
top_region = region_analysis.iloc[0]["region"]

print("Top Region:", top_region)

# 25. Top Category
top_category = category_analysis.iloc[0]["category"]

print("Top Category:", top_category)

# 26. Top Customer Segment
top_segment = segment_analysis.iloc[0]["customer_segment"]

print("Top Customer Segment:", top_segment)

# ==========================================
# FINAL SUMMARY
# ==========================================

summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Total Orders",
        "Average Sales",
        "Average Profit",
        "Profit Margin",
        "Top Product",
        "Top Region",
        "Top Category",
        "Top Customer Segment"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        total_orders,
        average_sales,
        average_profit,
        round(profit_margin, 2),
        top_product,
        top_region,
        top_category,
        top_segment
    ]
})

print("\nFinal Summary:")
display(summary)

# ==========================================
# ONE OUTPUT FILE ONLY
# ==========================================

output_file = "Python_Analysis_Result.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    summary.to_excel(
        writer,
        sheet_name="Summary",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region_Analysis",
        index=False
    )

    category_analysis.to_excel(
        writer,
        sheet_name="Category_Analysis",
        index=False
    )

    product_analysis.to_excel(
        writer,
        sheet_name="Product_Analysis",
        index=False
    )

    segment_analysis.to_excel(
        writer,
        sheet_name="Segment_Analysis",
        index=False
    )

    monthly_sales.to_excel(
        writer,
        sheet_name="Monthly_Sales",
        index=False
    )

    monthly_profit.to_excel(
        writer,
        sheet_name="Monthly_Profit",
        index=False
    )

    yearly_sales.to_excel(
        writer,
        sheet_name="Yearly_Sales",
        index=False
    )

    yearly_profit.to_excel(
        writer,
        sheet_name="Yearly_Profit",
        index=False
    )

    region_category.to_excel(
        writer,
        sheet_name="Region_Category",
        index=False
    )

print("\nAnalysis Completed Successfully!")
print("Only one output file created:")
print(output_file)

files.download(output_file)
```
